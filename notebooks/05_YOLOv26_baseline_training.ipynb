```python
from ultralytics import YOLO
from pathlib import Path
import torch
import os
```

```python
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
PROJECT_ROOT = Path(
    r"D:\master\Master_Drone_Detection"
)


DATA_YAML = (
    PROJECT_ROOT
    /
    "datasets"
    /
    "AntiUAV_YOLO"
    /
    "data.yaml"
)


print(DATA_YAML)
print(DATA_YAML.exists())
```

```python
EXP_DIR = (
    PROJECT_ROOT
    /
    "experiments"
    /
    "EXP001_YOLOv26_baseline"
)


EXP_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print(EXP_DIR)
```

```python
MODEL_NAME = "yolo26s.pt"


model = YOLO(MODEL_NAME)
```

```python
results = model.train(

    data=str(DATA_YAML),

    epochs=100,

    imgsz=960,

    batch=4,

    device=0,

    workers=2,

    project=str(EXP_DIR),

    name="baseline",

    pretrained=True,

    optimizer="auto",

    patience=30,

    save=True,

    plots=True

)
```
