# EXP002: YOLOv8s High-Resolution Training

Training the YOLOv8s baseline at 960px image resolution.

```python
from pathlib import Path
from ultralytics import YOLO

PROJECT_PATH = Path("..")
EXP2_DIR = PROJECT_PATH / "04_experiments" / "EXP002_YOLOv8s_high_resolution"
DATA_YAML = PROJECT_PATH / "02_datasets" / "DUT_Anti_UAV" / "data.yaml"
MODEL_PATH = PROJECT_PATH / "05_models" / "pretrained" / "yolov8s.pt"

EXP2_DIR.mkdir(parents=True, exist_ok=True)

model = YOLO(str(MODEL_PATH))

results = model.train(
    data=str(DATA_YAML),
    epochs=100,
    imgsz=960,
    batch=4,
    device=0,
    project=str(EXP2_DIR),
    name="YOLOv8s_960",
    pretrained=True,
    patience=20,
    workers=4,
)

print("Training completed.")
print("Results directory:", EXP2_DIR / "YOLOv8s_960")
```

```python
from ultralytics import YOLO 

exp2_model = (
    "runs/04_experiments/"
    "EXP002_YOLOv8s_high_resolution/"
    "YOLOv8s_960-3/weights/best.pt"
             )          # saved the trained model 
model_exp2 = YOLO(exp2_model) #load the model in memory 
```

```python
metrics_exp2 = model_exp2.val(
    data="D:/master/Master_Drone_Detection/02_datasets/DUT_Anti_UAV/data.yaml",
    imgsz=960,
    device=0,
)

summary = metrics_exp2.results_dict
print("EXP002 validation metrics")
print(f"Precision: {summary['metrics/precision(B)']:.4f}")
print(f"Recall:    {summary['metrics/recall(B)']:.4f}")
print(f"mAP50:     {summary['metrics/mAP50(B)']:.4f}")
print(f"mAP50-95:  {summary['metrics/mAP50-95(B)']:.4f}")
print("Results directory:", metrics_exp2.save_dir)
```
