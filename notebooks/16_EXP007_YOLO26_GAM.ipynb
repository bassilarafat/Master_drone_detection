```python
from ultralytics import YOLO
from pathlib import Path

PROJECT = Path(r"D:\master\Master_Drone_Detection")

DATA_YAML = PROJECT / "02_datasets" / "DUT_Anti_UAV" / "data.yaml"

model = YOLO(
    str(PROJECT / "05_models" / "pretrained" / "yolo26s_gam.yaml")
)

model.train(
    data=str(DATA_YAML),
    epochs=100,
    imgsz=960,
    batch=4,
    workers=2,
    device=0,
    project=str(PROJECT / "04_experiments"),
    name="YOLO26s_GAM",
    pretrained=str(
        PROJECT /
        "04_experiments" /
        "YOLO26s_baseline-5" /
        "weights" /
        "best.pt"
    ),
    cache=False
)
```

```python
from ultralytics import YOLO

best_model = YOLO(r"D:\master\Master_Drone_Detection\04_experiments\YOLO26s_GAM\weights\best.pt")

metrics = best_model.val(
    data=r"D:\master\Master_Drone_Detection\02_datasets\DUT_Anti_UAV\data.yaml",
    imgsz=960,
    batch=4,
    device=0
)

print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```
