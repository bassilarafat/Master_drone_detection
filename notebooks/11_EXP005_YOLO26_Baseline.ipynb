```python
from ultralytics import YOLO

model = YOLO("yolo26s.pt")

print("YOLO26 loaded")
```

```python
from pathlib import Path


PROJECT = Path(
    r"D:\master\Master_Drone_Detection"
)


DATASET = (
    PROJECT /
    "02_datasets" /
    "DUT_Anti_UAV"
)


DATA_YAML = DATASET / "data.yaml"


print(DATA_YAML)
```

```python
EXP005_DIR = PROJECT / "04_experiments"

model.train(
    data=str(DATA_YAML),
    epochs=100,
    imgsz=960,
    batch=4,
    workers=2,
    device=0,
    project=str(EXP005_DIR),
    name="YOLO26s_baseline",
)
```

```python
from ultralytics import YOLO

model = YOLO(
    r"D:\master\Master_Drone_Detection\04_experiments\YOLO26s_baseline-5\weights\best.pt"
)

metrics = model.val(
    data=r"D:\master\Master_Drone_Detection\02_datasets\DUT_Anti_UAV\data.yaml",
    split="test",
    imgsz=960,
    batch=2,
    device=0
)

print(metrics)
```
